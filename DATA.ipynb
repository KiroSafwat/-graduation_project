### Protocol Analysis

**Conclusion:** TCP contains both benign traffic and multiple attack types, while UDP traffic in this dataset is almost entirely labeled BENIGN. Protocol 0 represents a very small portion of the dataset and requires further investigation. Therefore, Protocol may be a useful feature for distinguishing traffic patterns, but its predictive value must be evaluated during model training and testing.

### Flow Duration Analysis

The Flow Duration feature ranges from -13 to 120,000,000, with a median of 31,328. A total of 115 flows have negative duration values, all of which are labeled BENIGN. Since flow duration represents the elapsed time of a network flow, negative values are invalid and should be investigated or handled during preprocessing.

There are also 1,506 flows with a duration of zero. These include 1,365 BENIGN flows, 126 PortScan flows, 10 Bot flows, 3 FTP-Patator flows, and 2 DDoS flows. Therefore, zero-duration flows should not be removed automatically, as they occur in both benign and attack traffic and may contain useful information.

**Conclusion:** Negative duration values are invalid and require preprocessing, while zero-duration values should be retained unless further investigation shows that they are measurement artifacts.

### Zero Values in Total Backward Packets

A total of 452,159 flows (approximately 15.98% of the dataset) have zero recorded backward packets. Of these, 330,542 are labeled BENIGN, 66,110 are DoS Hulk, and 46,528 are DDoS. Zero backward-packet counts are therefore present in both benign and attack traffic.

These records should not be treated as invalid solely because their backward-packet count is zero. Further analysis of the proportion of zero-backward flows within each traffic label is needed to determine whether this feature helps distinguish benign traffic from specific attack categories.

```python
zero_backward = (
    df[" Total Backward Packets"] == 0
)

zero_rate_by_label = (
    df.assign(Zero_Backward=zero_backward)
      .groupby(" Label")["Zero_Backward"]
      .mean()
      .mul(100)
      .sort_values(ascending=False)
)

print(zero_rate_by_label.round(2))
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    y=zero_rate_by_label.values,
    x=zero_rate_by_label.index
)

```

### Relationship Between Zero Backward Packets and Traffic Labels

The proportion of flows with zero backward packets varies considerably across traffic labels. DoS Slowhttptest has the highest proportion at 57.68%, followed by DDoS at 36.34%, DoS Hulk at 28.73%, and DoS slowloris at 28.33%. In comparison, 14.55% of BENIGN flows have zero backward packets, while the proportion for PortScan is only 0.04%.

These differences suggest that zero backward-packet counts may be associated with certain traffic patterns, particularly some DoS attacks, in this dataset. However, zero values also occur in benign traffic, so this feature should not be used alone to classify traffic. The observed relationships should be evaluated further during model training and testing.

### Comparison Between Forward and Backward Packet Lengths

The comparison reveals distinct zero-value patterns across different traffic labels. Several attack categories, including Web Attack–XSS, Web Attack–Brute Force, SQL Injection, and DoS GoldenEye, exhibit identical zero percentages in both directions.

In contrast, some categories show substantially higher zero percentages in the backward direction. FTP-Patator has a backward zero rate of 49.95%, compared to only 0.23% in the forward direction. Similar patterns appear in DoS slowloris, DDoS, and DoS Slowhttptest.

Conversely, PortScan and Bot exhibit considerably higher zero percentages in the forward direction than in the backward direction.

These differences indicate that zero packet-length values are distributed differently across traffic categories and directions. Therefore, analyzing forward and backward packet lengths separately may provide useful information for distinguishing traffic patterns. However, zero values alone should not be treated as sufficient evidence for classifying an attack.
